## Step 6: Load Models

This notebook uses:

- **Grounding DINO** for saree detection
- **SAM** for segmentation
- **Stable Diffusion 2 Inpainting** for final image generation

```python
GROUNDING_DINO_MODEL_ID = "IDEA-Research/grounding-dino-base"
INPAINT_MODEL_ID = "sd2-community/stable-diffusion-2-inpainting"

processor = AutoProcessor.from_pretrained(GROUNDING_DINO_MODEL_ID)
grounding_model = AutoModelForZeroShotObjectDetection.from_pretrained(
    GROUNDING_DINO_MODEL_ID
).to(DEVICE)

sam = sam_model_registry["vit_b"](checkpoint=SAM_CHECKPOINT).to(device=DEVICE)
sam_predictor = SamPredictor(sam)

pipe = StableDiffusionInpaintPipeline.from_pretrained(
    INPAINT_MODEL_ID,
    torch_dtype=DTYPE,
)
pipe = pipe.to(DEVICE)

if DEVICE == "cuda":
    pipe.enable_attention_slicing()

print("All models loaded successfully.")
```

## Step 7: Define Helper Functions

These helper functions are used for image display, saree box detection, mask generation, and refinement of segmentation masks.

```python
def show_image(img, title="", figsize=(6, 6), cmap=None):
    plt.figure(figsize=figsize)
    plt.imshow(img, cmap=cmap)
    plt.title(title)
    plt.axis("off")
    plt.show()

def detect_best_box(image_pil, candidate_labels=("saree",), box_threshold=0.25, text_threshold=0.20):
    best = None
    H, W = image_pil.size[1], image_pil.size[0]

    for label in candidate_labels:
        text_prompt = f"{label}."
        inputs = processor(images=image_pil, text=text_prompt, return_tensors="pt").to(DEVICE)
        with torch.no_grad():
            outputs = grounding_model(**inputs)

        results = processor.post_process_grounded_object_detection(
            outputs, inputs.input_ids, threshold=box_threshold, text_threshold=text_threshold, target_sizes=[(H, W)]
        )[0]

        boxes = results["boxes"].detach().cpu().numpy() if len(results["boxes"]) > 0 else np.array([])
        scores = results["scores"].detach().cpu().numpy() if len(results["scores"]) > 0 else np.array([])

        if len(boxes) == 0: continue

        for box, score in zip(boxes, scores):
            x1, y1, x2, y2 = box
            area = max(1.0, (x2 - x1) * (y2 - y1))
            rank_score = float(score) * float(area)
            item = {"label": label, "box": box.astype(np.float32), "score": float(score), "rank_score": rank_score}
            if best is None or item["rank_score"] > best["rank_score"]:
                best = item

    if best is None:
        raise ValueError("No region detected for labels: " + str(candidate_labels))
    return best

def segment_from_box(image_np, box_xyxy):
    sam_predictor.set_image(image_np)
    masks, scores, _ = sam_predictor.predict(box=box_xyxy, multimask_output=True)
    best_idx = int(np.argmax(scores))
    return masks[best_idx].astype(np.uint8) * 255

def refine_preserve_mask(mask_uint8, dilate_kernel=9):
    kernel = np.ones((dilate_kernel, dilate_kernel), np.uint8)
    refined = cv2.morphologyEx(mask_uint8, cv2.MORPH_CLOSE, kernel)
    refined = cv2.dilate(refined, kernel, iterations=1)
    return refined

def make_inpaint_mask_from_preserve_mask(preserve_mask_uint8):
    return 255 - preserve_mask_uint8

def seamless_feature_blend(original_np, ai_generated_np, guard_mask_np):
    """Physically grafts the original face/hands back onto the AI body perfectly."""
    blended_np = ai_generated_np.copy()
    contours, _ = cv2.findContours(guard_mask_np, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

    for cnt in contours:
        x, y, w, h = cv2.boundingRect(cnt)
        if w < 10 or h < 10: continue

        center = (x + w // 2, y + h // 2)
        local_mask = np.zeros_like(guard_mask_np)
        cv2.drawContours(local_mask, [cnt], -1, 255, -1)

        try:
            blended_np = cv2.seamlessClone(original_np, blended_np, local_mask, center, cv2.NORMAL_CLONE)
        except cv2.error:
            blended_np[local_mask > 127] = original_np[local_mask > 127]
    return blended_np
```

## Step 8: Detect the Saree Region

Grounding DINO is used to find the best saree-related bounding box.

```python
def get_saree_bounding_box(source_pil, source_np, thresholds=(0.25, 0.20)):
    """
    Detects the best saree bounding box using Grounding DINO,
    visualizes the result, and returns the bounding box coordinates.
    """
    print("Detecting saree region...")

    # Run the detection
    best_detection = detect_best_box(
        image_pil=source_pil,
        candidate_labels=("saree", "indian saree", "traditional saree", "garment", "dress"),
        box_threshold=thresholds[0],
        text_threshold=thresholds[1]
    )

    # Extract required data
    best_box = best_detection["box"]
    best_label = best_detection["label"]
    best_score = best_detection["score"]

    # Print diagnostics
    print(f"✓ Detection Complete.")
    print(f"  - Label: {best_label}")
    print(f"  - Score: {best_score:.3f}")
    print(f"  - Box: {best_box}")

    # Visualize the bounding box
    boxed_image = overlay_box(source_np, best_box, f"{best_label}: {best_score:.3f}")
    show_image(boxed_image, "Grounding DINO Detection", figsize=(6, 8))

    # Return the exact variable needed for SAM segmentation
    return best_box
```

## Step 10: Save Intermediate Masks

```python
def generate_and_display_masks(source_np, best_box, show_visuals=True):
    """
    Segments the target object using SAM, refines the mask, creates the
    inpainting mask, and optionally displays the steps.

    Returns:
        preserve_mask (numpy.ndarray): The black-and-white mask of the protected item.
        inpaint_mask (numpy.ndarray): The inverted mask where White = AI Redraw Zone.
    """
    print("Generating pixel-perfect masks using SAM...")

    # 1. Get raw segmentation from SAM using the DINO bounding box
    raw_preserve_mask = segment_from_box(source_np, best_box)

    # 2. Clean up the edges (closes gaps and expands slightly)
    preserve_mask = refine_preserve_mask(raw_preserve_mask, dilate_kernel=9)

    # 3. Create the Inpaint Mask (Invert: White = AI can draw, Black = Locked)
    inpaint_mask = make_inpaint_mask_from_preserve_mask(preserve_mask)

    # 4. Display the diagnostic visuals
    if show_visuals:
        show_image(raw_preserve_mask, "Raw Preserve Mask", figsize=(6, 8), cmap="gray")
        show_image(preserve_mask, "Refined Preserve Mask", figsize=(6, 8), cmap="gray")
        show_image(inpaint_mask, "Final Inpaint Mask (white = repaint)", figsize=(6, 8), cmap="gray")

    print("✓ Mask generation complete.")
    return preserve_mask, inpaint_mask
```

```python
PROMPT = """
Ultra-realistic full-body beautiful extremely heavyset plus-sized Indian woman standing naturally,
wearing the same exact saree from the original image.
Preserve the saree fabric, texture, print, borders, folds, drape, and colors accurately.
Natural pose, wide waist, broad shoulders, clean background, studio-quality lighting.
"""

NEGATIVE_PROMPT = """
low quality, blurry, deformed, distorted body, extra arms, extra fingers,
missing fingers, bad hands, bad anatomy, duplicate body, skinny, thin,
unrealistic saree, changed saree pattern, washed fabric details
"""

SEED = 1234
NUM_STEPS = 35
GUIDANCE_SCALE = 8.5
```

## Step 12: Run Stable Diffusion Inpainting

The white region of the inpainting mask is regenerated while the saree region is preserved.

```python
def generate_final_image(pipe, source_pil, inpaint_mask_np, prompt, negative_prompt, seed, num_steps, guidance_scale, strength, device):
    generator = torch.Generator(device=device).manual_seed(seed)
    mask_pil = Image.fromarray(inpaint_mask_np).convert("L").resize(source_pil.size)

    result = pipe(
        prompt=prompt,
        negative_prompt=negative_prompt,
        image=source_pil,
        mask_image=mask_pil,
        guidance_scale=guidance_scale,
        num_inference_steps=num_steps,
        strength=strength,
        generator=generator,
    ).images[0]

    # Resize output back to absolute original dimensions to prevent alignment errors
    return result.resize(source_pil.size, Image.LANCZOS)
```

 ## Step 14: Download Final Files


```python
import os
def download_generated_files(file_paths):
    """
    Triggers a browser download for a list of file paths in Google Colab.
    Returns a list of files that were successfully queued for download.
    """
    try:
        from google.colab import files
    except ImportError:
        print("❌ google.colab module not found. File downloading is only supported in Google Colab.")
        return []

    print(f"Preparing to download {len(file_paths)} files...")
    downloaded_files = []

    for path in file_paths:
        if os.path.exists(path):
            print(f"  -> Downloading: {os.path.basename(path)}")
            files.download(path)
            downloaded_files.append(path)
        else:
            print(f"  -> ⚠️ Warning: Skipped {os.path.basename(path)} (File not found)")

    print(f"✓ Successfully queued {len(downloaded_files)} files for download.")
    return downloaded_files
```

```python
INPUT_FOLDER = "/content"
# Strictly grab Image_1.png to Image_8.png
image_files = [os.path.join(INPUT_FOLDER, f"Image_{i}.png") for i in range(1, 9)]
image_files = [f for f in image_files if os.path.exists(f)]

if not image_files:
    print("⚠️ No target images found! Please ensure Image_1.png to Image_8.png are uploaded.")
else:
    print(f"🚀 Found {len(image_files)} target images. Starting Batch Processing...\n")

    fig, axes = plt.subplots(len(image_files), 3, figsize=(15, 6 * len(image_files)))
    if len(image_files) == 1: axes = [axes]

    for i, img_path in enumerate(image_files):
        file_name = os.path.basename(img_path)
        print(f"--- Processing [{i+1}/{len(image_files)}]: {file_name} ---")

        # --- A. Load Image ---
        source_pil = Image.open(img_path).convert("RGB")
        source_np = np.array(source_pil)

        # --- B. Smart Masking (Protect Face & Hands ONLY) ---
        try:
            face_box = detect_best_box(source_pil, candidate_labels=("human face", "head"), box_threshold=0.20)["box"]
            face_mask = segment_from_box(source_np, face_box)
        except Exception:
            face_mask = np.zeros((source_np.shape[0], source_np.shape[1]), dtype=np.uint8)

        try:
            hands_box = detect_best_box(source_pil, candidate_labels=("human hands", "fingers"), box_threshold=0.15)["box"]
            hands_mask = segment_from_box(source_np, hands_box)
        except Exception:
            hands_mask = np.zeros((source_np.shape[0], source_np.shape[1]), dtype=np.uint8)

        # Combine into Master Guard and refine
        raw_preserve = cv2.bitwise_or(face_mask, hands_mask)
        preserve_mask = refine_preserve_mask(raw_preserve, dilate_kernel=15)
        inpaint_mask = make_inpaint_mask_from_preserve_mask(preserve_mask)

        # --- C. Generate AI Reshaping ---
        raw_sd_pil = generate_final_image(
            pipe=pipe, source_pil=source_pil, inpaint_mask_np=inpaint_mask,
            prompt=PROMPT, negative_prompt=NEGATIVE_PROMPT,
            seed=SEED, num_steps=NUM_STEPS, guidance_scale=GUIDANCE_SCALE,
            strength=0.65, device=DEVICE
        )

        # --- D. Poisson Seamless Cloning (THE FIX FOR BAD FACES) ---
        raw_sd_np = np.array(raw_sd_pil)
        # We physically graft the pristine original face/hands over the AI generation
        blended_np = seamless_feature_blend(source_np, raw_sd_np, preserve_mask)
        final_pil = Image.fromarray(blended_np)

        # --- E. Save Output ---
        save_path = os.path.join(INPUT_FOLDER, f"Fat_{file_name}")
        # Ensure the function is defined or define it locally as a fallback
        if 'save_final_output' in globals():
            save_final_output(final_pil, save_path)
        else:
            final_pil.save(save_path)

        # --- F. Plot Data to Grid ---
        axes[i][0].imshow(source_pil)
        axes[i][0].set_title(f"Original {i+1}", fontsize=14, fontweight="bold")
        axes[i][0].axis("off")

        axes[i][1].imshow(inpaint_mask, cmap='gray')
        axes[i][1].set_title(f"AI Redraw Zone {i+1}\n(Face/Hands Locked)", fontsize=12)
        axes[i][1].axis("off")

        axes[i][2].imshow(final_pil)
        axes[i][2].set_title(f"Fat Output {i+1}", fontsize=14, fontweight="bold", color="green")
        axes[i][2].axis("off")

        # --- G. VRAM Memory Cleanup ---
        gc.collect()
        if torch.cuda.is_available(): torch.cuda.empty_cache()

    # Render Final Grid
    plt.tight_layout()
    plt.show()

    fig.savefig("/content/full_results_grid.png", bbox_inches='tight')
    print("\n✅ Batch Processing Complete! Grid saved as 'full_results_grid.png'")
```
